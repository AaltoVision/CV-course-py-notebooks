```python
def gaussian_kernel_1d(sigma: float, truncate: float = 4.0) -> np.ndarray:
    """
    Create a 1D Gaussian kernel with dynamic size based on sigma.
    
    Args:
        sigma (float): Standard deviation of the Gaussian.
        truncate (float): Truncation value for kernel size computation.
    
    Returns:
        np.ndarray: 1D Gaussian kernel normalized to sum to 1.
    """
    # Calculate dynamic kernel size
    radius = int(truncate * sigma + 0.5)

    # Create Gaussian kernel
    x = np.arange(-radius, radius + 1)
    kernel = np.exp(-(x**2) / (2 * sigma**2))
    kernel /= kernel.sum()
    return kernel


def apply_gaussian_filter_separable(image: np.ndarray, sigma: float, truncate: float = 4.0) -> np.ndarray:
    """
    Apply a separable 2D Gaussian filter using horizontal and vertical 1D convolutions.
    
    Args:
        image (np.ndarray): Input image, a 2D (grayscale) or 3D (color) NumPy array.
        sigma (float): Standard deviation of the Gaussian.
        truncate (float): Truncation value for kernel size computation.
    
    Returns:
        np.ndarray: Filtered image with the same shape as the input.
    """
    # Generate the 1D Gaussian kernel
    kernel = gaussian_kernel_1d(sigma, truncate)

    # Apply horizontal and vertical 1D convolutions
    temp = convolve1d(image, kernel, axis=1, mode='reflect')  # Horizontal pass
    filtered_image = convolve1d(temp, kernel, axis=0, mode='reflect')  # Vertical pass
    return filtered_image
```

```python
def median_filter_c(image: np.ndarray, size: int = 5, mode: str = "symmetric") -> np.ndarray:
    """
    Apply a 2D median filter to an image manually (without external libraries).
    
    Args:
        image (np.ndarray): Input image, a 2D (grayscale) NumPy array.
        size (int): Neighborhood size for the median filter (default is 5x5).
    
    Returns:
        np.ndarray: Filtered image with the same shape as the input.
    """
    if size <= 0 or size % 2 == 0:
        raise ValueError("Size must be a positive odd integer.")
    if image.ndim != 2:
        raise ValueError("Only 2D grayscale images are supported.")

    # Padding size
    pad_size = size // 2
    
    # Pad the image to handle borders
    padded_image = np.pad(image, pad_width=pad_size, mode=mode)
    
    # Create an empty array for the filtered image
    filtered_image = np.zeros_like(image)
    
    # Iterate over each pixel in the image
    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            # Extract the neighborhood
            neighborhood = padded_image[i:i + size, j:j + size]
            # Compute the median and assign it to the pixel
            filtered_image[i, j] = np.median(neighborhood)
    
    return filtered_image
```

```python
def bilateral_filter_c(image: np.ndarray, wsize: int, sigma_d: float, sigma_r: float) -> np.ndarray:
    """
    Apply a bilateral filter to an image.
    
    Args:
        image (np.ndarray): Input image, a 2D (grayscale) NumPy array with values in the range [0, 1].
        wsize (int): The size (diameter) of the neighborhood for filtering.
        sigma_d (float): Sigma value for domain kernel.
        sigma_r (float): Sigma value for range kernel.
    
    Returns:
        np.ndarray: Filtered image with the same shape as the input.
    """
    if image.ndim != 2:
        raise ValueError("Only 2D grayscale images are supported.")
    
    # Pad the image to handle borders
    pad_size = wsize // 2
    padded_image = np.pad(image, pad_width=pad_size, mode='reflect')
    
    # Create an empty array for the filtered image
    filtered_image = np.zeros_like(image)
    
    # Precompute the spatial Gaussian kernel
    x, y = np.meshgrid(np.arange(-pad_size, pad_size + 1), np.arange(-pad_size, pad_size + 1))
    domain_kernel = np.exp(-(x**2 + y**2) / (2 * sigma_d**2))
    
    # Iterate over each pixel in the image
    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            # Extract the neighborhood
            neighborhood = padded_image[i:i + wsize, j:j + wsize]
            
            # Compute the intensity Gaussian kernel
            range_kernel = np.exp(-((neighborhood - image[i, j])**2) / (2 * sigma_r**2))
            
            # Compute the combined bilateral kernel
            bilateral_kernel = domain_kernel * range_kernel
            
            # Normalize the kernel
            bilateral_kernel /= bilateral_kernel.sum()
            
            # Compute the filtered pixel value
            filtered_image[i, j] = np.sum(bilateral_kernel * neighborhood)
    
    return filtered_image
```

```python
def gaussian_filter_c(shape, cutoff, high_pass=False):
    rows, cols = shape
    crow, ccol = rows // 2, cols // 2
    x, y = np.ogrid[:rows, :cols]
    mask = np.exp(-((x - crow)**2 + (y - ccol)**2) / (2 * (cutoff**2)))
    return 1 - mask if high_pass else mask
```
